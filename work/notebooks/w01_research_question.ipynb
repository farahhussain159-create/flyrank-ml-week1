## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/farahhussain159-create/flyrank-ml-week1/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print("Total rows:", len(df))
print("Total columns:", len(df.columns))

# Derive the declining label from trend_direction
df['is_declining_label'] = (df['trend_direction'] == 'down').astype(int)

declining_pct = round(df['is_declining_label'].mean() * 100, 2)
print("Declining %:", declining_pct, "%")

print("\nTrend direction breakdown:")
print(df['trend_direction'].value_counts())
```

A quick look at this working dataset shows 30,000 rows across 44 columns, drawn from the FlyRank ML Internship dataset. Of these, 54.21% of pages are labeled as declining (derived from trend_direction == "down"), with the remaining split across stable, up, new, and flat trends. This scale and the near-balanced split between declining and non-declining pages make the lane worth pursuing over the next 7 weeks — there's enough signal in both classes to train and validate a meaningful classifier.
